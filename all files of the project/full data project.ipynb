```python
import numpy as np
import pandas as pd

file_numbers = [str(i).zfill(2) for i in range(1, 13)]

df_list = []

for num in file_numbers:
    file_path = f'/content/Orders_{num}.csv'
    try:
        df = pd.read_csv(file_path)
        df['Month'] = num
        df_list.append(df)
    except FileNotFoundError:
        print(f"Warning: File not found - {file_path}. Skipping.")

df_orders = pd.concat(df_list, ignore_index=True)
```

```python
df_orders.info()
```

```python
df_orders[df_orders['Quantity'].isnull()]
```

```python
df_cleaned = df_orders.dropna()
df_cleaned.info()
```

```python
df_cleaned['OrderDate']
```

```python
df_cleaned['OrderDate'] = df_cleaned['OrderDate'].str.replace('-', '/')

df_cleaned['split_date'] = df_cleaned['OrderDate'].str.split('/')
df_cleaned
df_cleaned['date1'] = df_cleaned['split_date'].str[0]
df_cleaned['date2'] = df_cleaned['split_date'].str[1]
df_cleaned['date3'] = df_cleaned['split_date'].str[2]
df_cleaned.drop(columns=['split_date'], inplace=True)
df_cleaned.info()
```

```python
df_cleaned.info()
```

```python
df_cleaned['Year'] = None
df_cleaned['Day'] = None
```

```python
df_cleaned['date1']=df_cleaned['date1'].astype(int)
df_cleaned['date2']=df_cleaned['date2'].astype(int)
df_cleaned['date3']=df_cleaned['date3'].astype(int)
df_cleaned['Month']=df_cleaned['Month'].astype(int)
df_cleaned.info()
```

```python
cond1_day = (df_cleaned['date1']>=1) & (df_cleaned['date1']<=31) & (df_cleaned['date1'] != df_cleaned['Month'])
cond2_day = (df_cleaned['date2']>=1) & (df_cleaned['date2']<=31) & (df_cleaned['date2'] != df_cleaned['Month'])
cond3_day = (df_cleaned['date3']>=1) & (df_cleaned['date3']<=31) & (df_cleaned['date3'] != df_cleaned['Month'])
df_cleaned.loc[cond1_day, 'Day'] = df_cleaned.loc[cond1_day, 'date1']
df_cleaned.loc[cond2_day, 'Day'] = df_cleaned.loc[cond2_day, 'date2']
df_cleaned.loc[cond3_day, 'Day'] = df_cleaned.loc[cond3_day, 'date3']
df_cleaned.loc[df_cleaned['date1']>1000, 'Year'] = df_cleaned.loc[df_cleaned['date1']>1000, 'date1']
df_cleaned.loc[df_cleaned['date2']>1000, 'Year'] = df_cleaned.loc[df_cleaned['date2']>1000, 'date2']
df_cleaned.loc[df_cleaned['date3']>1000, 'Year'] = df_cleaned.loc[df_cleaned['date3']>1000, 'date3']
df_cleaned.info()
```

```python
df_cleaned['Day'].fillna(df_cleaned['Month'], inplace=True)
df_cleaned.info()
```

```python
df_cleaned.drop(columns=['date1', 'date2', 'date3'], inplace=True)
df_cleaned.info()
```

```python
df_cleaned['OrderID'] = df_cleaned['OrderID'].astype(str)
df_cleaned['CustomerID'] = df_cleaned['CustomerID'].astype(str)
df_cleaned['ProductID'] = df_cleaned['ProductID'].astype(str)
df_cleaned.info()
```

```python
df_cleaned['CleanedDate'] = pd.to_datetime(df_cleaned['Year'].astype(str) + '/' + df_cleaned['Month'].astype(str) + '/' + df_cleaned['Day'].astype(str))
df_cleaned
```

```python
df_cleaned.to_csv('full_orders_cleaned')
```
